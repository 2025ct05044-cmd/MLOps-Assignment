```python
columns=['age','sex','cp','trestbps','chol','fbs','restecg','thalach','exang','oldpeak','slope','ca','thal','target']
df=pd.read_csv('HeartDisease_MLOps/data/raw/heart_disease.csv',names=columns)
print(df.shape)
df.head()
```

```python
df.replace('?',np.nan,inplace=True)
df['ca']=pd.to_numeric(df['ca'])
df['thal']=pd.to_numeric(df['thal'])
df.dropna(inplace=True)
df['target']=(df['target']>0).astype(int)
print(df.shape)
```

```python
processed_destination='HeartDisease_MLOps/data/processed/cleaned_heart_disease.csv'
df.to_csv(processed_destination,index=False)
print('saved',processed_destination)
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['age'],bins=20,kde=True)
plt.title('Age Distribution')
plt.savefig('HeartDisease_MLOps/screenshots/age_distribution.png',dpi=300)
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='target',data=df)
plt.title('Heart Disease Distribution')
plt.savefig('HeartDisease_MLOps/screenshots/class_balance.png',dpi=300)
plt.show()
```

```python
plt.figure(figsize=(12,8))
sns.heatmap(df.corr(),annot=True,cmap='coolwarm',fmt='.2f')
plt.title('Correlation Heatmap')
plt.savefig('HeartDisease_MLOps/screenshots/correlation_heatmap.png',dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='target',y='chol',data=df)
plt.title('Cholesterol by Disease Status')
plt.savefig('HeartDisease_MLOps/screenshots/cholesterol_boxplot.png',dpi=300)
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='target',y='age',data=df)
plt.title('Age by Disease Status')
plt.savefig('HeartDisease_MLOps/screenshots/age_disease_boxplot.png',dpi=300)
plt.show()
```
